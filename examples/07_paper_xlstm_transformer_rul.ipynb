# Paper Reproduction: xLSTM-Transformer for Bearing RUL

This notebook reproduces a Jiang et al. style xLSTM-Transformer RUL workflow inside the project framework. It trains an xLSTM-inspired Transformer and two feature-sequence baselines on XJTU-SY and PHM2012 style bearing splits. The notebook is a smoke executable example; formal 50-epoch real-data evidence is produced by scripts/run_formal_paper_reproductions.py.

```python
from pathlib import Path

import pandas as pd

from USTC.SSE.BearingPrediction.examples import run_paper_xlstm_transformer_reproduction
```

```python
result = run_paper_xlstm_transformer_reproduction(prefer_real_data=True, max_samples_per_entity=16)
comparison_path = Path(result["comparison_path"])
assert comparison_path.name == "comparison_metrics.csv"
comparison_frame = pd.read_csv(comparison_path)
required_columns = {
    "rmse",
    "normalized_rmse",
    "r2",
    "r2_score",
    "phm2012_score",
    "huang_rul_score",
    "rmse_change_pct_vs_transformer",
    "score_change_pct_vs_transformer",
}
assert required_columns.issubset(comparison_frame.columns)
assert "XLSTM-Transformer" in set(comparison_frame["model_name"])
comparison_frame
```
